## Retrieve Detailed Schema for Public Tables

### Subtask:
Query the `information_schema.columns` to fetch detailed schema information (column names, data types, etc.) for all tables that belong to the `public` schema.


**Reasoning**:
To retrieve detailed schema information for public tables, I will use the `%%sql` magic command to query `information_schema.columns` and filter for the 'public' schema, then order the results for clarity.



````python
%%sql

SELECT
    *
FROM
    information_schema.columns
WHERE
    table_schema = 'public'
ORDER BY
    table_name, ordinal_position;
````

### Organize and Present Schema Details

To make the schema details more understandable, I will process the DataFrame obtained from the previous SQL query (`_`). I'll group the results by `table_name` and then iterate through each table to present its `column_name`, `data_type`, `character_maximum_length` (if applicable), and `is_nullable` status. This structured output will provide a clear overview of each table's structure.

**Reasoning**:
Now, I will implement the logic to organize and present the schema details. I'll group the DataFrame by `table_name` and then iterate through each table, printing its name and then a formatted list of its columns, their data types, max lengths (if applicable), and nullability.



````python
schema_df = _ # The result of the last SQL query is stored in the '_' variable

for table_name, group in schema_df.groupby('table_name'):
    print(f"\n--- Table: {table_name} ---")
    print(f"{'Column Name':<30} {'Data Type':<20} {'Max Length':<15} {'Nullable'}")
    print(f"{'='*30} {'='*20} {'='*15} {'='*10}")
    for _, row in group.iterrows():
        col_name = row['column_name']
        data_type = row['data_type']
        max_length = row['character_maximum_length'] if pd.notna(row['character_maximum_length']) else 'N/A'
        is_nullable = row['is_nullable']
        print(f"{col_name:<30} {data_type:<20} {str(max_length):<15} {is_nullable}")
````

````python
%%sql
SELECT
    p.color,
    SUM(CASE WHEN c.age BETWEEN 18 AND 25 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS revenue_18_25,
    SUM(CASE WHEN c.age BETWEEN 26 AND 35 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS revenue_26_35,
    SUM(CASE WHEN c.age BETWEEN 36 AND 45 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS revenue_36_45,
    SUM(CASE WHEN c.age > 45 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS revenue_46_plus
FROM
    sales s
    LEFT JOIN product p ON s.productkey = p.productkey
    LEFT JOIN customer c ON s.customerkey = c.customerkey
WHERE
    s.orderdate BETWEEN '2023-01-01' AND '2023-12-31'
GROUP BY
    p.color
ORDER BY
    p.color;
````

````python
%%sql
WITH revenue_size AS (
SELECT st.countryname,
SUM(CASE WHEN st.squaremeters < 1000 THEN sa.netprice * sa.quantity / sa.exchangerate ELSE 0 END) AS revenue_small_store,
SUM(CASE WHEN st.squaremeters < 1000 THEN st.squaremeters ELSE 0 END) AS total_sqm_small_store,
SUM(CASE WHEN st.squaremeters BETWEEN 1000 AND 2000 THEN sa.netprice * sa.quantity / sa.exchangerate ELSE 0 END) AS revenue_medium_store,
SUM(CASE WHEN st.squaremeters BETWEEN 1000 AND 2000 THEN st.squaremeters ELSE 0 END) AS total_sqm_medium_store,
SUM(CASE WHEN st.squaremeters > 2000 THEN sa.netprice * sa.quantity / sa.exchangerate ELSE 0 END) AS revenue_large_store,
SUM(CASE WHEN st.squaremeters > 2000 THEN st.squaremeters ELSE 0 END) AS total_sqm_large_store
FROM sales sa
LEFT JOIN store st ON sa.storekey = st.storekey
WHERE sa.orderdate BETWEEN '2023-01-01' AND '2023-12-31' AND st.countryname != 'Online'
GROUP BY 1
ORDER BY 1
)
SELECT countryname, SUM(revenue_small_store / NULLIF(total_sqm_small_store,0)) AS revenue_per_sqm_small_store,
SUM(revenue_medium_store / NULLIF(total_sqm_medium_store,0)) AS revenue_per_sqm_medium_store,
SUM(revenue_large_store / NULLIF(total_sqm_large_store,0)) AS revenue_per_sqm_large_store
FROM revenue_size
GROUP BY 1
ORDER BY 2 DESC, 3 DESC, 4 DESC
````

````python
# Store Revenue Segmentation (1.3.4)
%%sql
WITH store_revenue AS (
SELECT SUM(s.netprice * s.quantity / s.exchangerate) AS revenue,
st.storekey, st.squaremeters
FROM sales s
LEFT JOIN store st ON s.storekey = st.storekey
WHERE s.orderdate BETWEEN '2023-01-01' AND '2023-12-31'
GROUP BY 2)

SELECT
CASE WHEN squaremeters < 1000 AND revenue < 100000 THEN '1 - Small Store - Low Revenue'
WHEN squaremeters < 1000 AND revenue >= 100000 THEN '2 - Small Store - High Revenue'
WHEN squaremeters BETWEEN 1000 AND 2000 AND revenue < 300000 THEN '3 - Medium Store - Low Revenue'
WHEN squaremeters BETWEEN 1000 AND 2000 AND revenue >= 300000 THEN '4 - Medium Store - High Revenue'
WHEN squaremeters > 2000 AND revenue < 500000 THEN '5 - Large Store - Low Revenue'
WHEN squaremeters > 2000 AND revenue >= 500000 THEN '6 - Large Store - High Revenue'
ELSE '7 - Online Store' END AS store_category,
SUM(revenue) AS total_net_revenue,
SUM(revenue) / (SELECT SUM(revenue) FROM store_revenue) * 100 AS total_contribution
FROM store_revenue
GROUP BY 1
ORDER BY 1
````

````python
%%sql
WITH store_lifespan AS (SELECT *, EXTRACT(YEAR FROM AGE(closedate, opendate)) * 12 + EXTRACT(MONTH FROM AGE(closedate, opendate)) operational_lifespan
FROM store
WHERE status = 'Closed')

SELECT storekey, operational_lifespan,
CASE WHEN operational_lifespan < 12 * 5 THEN 'Less than 5 years'
WHEN operational_lifespan BETWEEN 12 * 5 AND 12 * 7 THEN '5-7 years'
WHEN operational_lifespan BETWEEN 7 * 5 AND 9 * 5 THEN '7-9 years'
ELSE '9+ years' END AS store_lifespan_category
FROM store_lifespan
ORDER BY 1
````

````python
%%sql
WITH customer_lifespan AS (SELECT *, EXTRACT(YEAR FROM AGE(enddt, startdt)) * 12 + EXTRACT(MONTH FROM AGE(enddt, startdt)) AS customer_lifespan
FROM customer
WHERE EXTRACT(YEAR FROM AGE(CURRENT_DATE,enddt)) >= 7)

SELECT
CASE WHEN customer_lifespan < 3 * 12 THEN '1 - Less than 3 years'
  WHEN customer_lifespan BETWEEN 3 * 12 AND 5 * 12 THEN '2 - 3-5 years'
  WHEN customer_lifespan BETWEEN 5 * 12 AND 7 * 12 THEN '3 - 5-7 years'
  ELSE '4 - 7+ years' END AS lifespan_category,
COUNT(customerkey) AS total_customers, # No need for distinct as customerkey is a key to the table. Hence, unqiue.
ROUND(COUNT(customerkey) * 100.0 / (SELECT COUNT(customerkey) FROM customer_lifespan),1) AS total_percent, # we actually can use 100.0 instead of casting float on one of the numbers. Also to get a total number we just had to count customerkeys from our CTE
ROUND(AVG(customer_lifespan),1) AS avergae_lifespan
FROM customer_lifespan
GROUP BY 1
ORDER BY 1
````

````python
# Customer LTV by Birth Decade (3.2.3) - Problem
# Calculate the lifetime value (LTV) for each individual customer and compare it to the average LTV of other customers in the same birth decade.
# This will help identify which customers—and which generations—tend to spend more over time.
# Comments on solution: we can join our data into one table that we can then use to perfrom calculation. We also had to group results to get one LTV per customerkey
%%sql
WITH ltv_table AS (
SELECT s.customerkey,
SUM(quantity * netprice / exchangerate) AS ltv,
EXTRACT(DECADE FROM c.birthday) * 10 AS birth_decade
FROM sales s
INNER JOIN customer c
ON s.customerkey = c.customerkey
GROUP BY 1, 3)

SELECT customerkey, birth_decade, ltv,
AVG(ltv) OVER(PARTITION BY birth_decade)
FROM ltv_table
WHERE ltv > 1000
ORDER BY 2,1
LIMIT 10
````

````python
# Revenue by Gender Analysis (3.2.4) - Problem
# Calculate the total net revenue generated by each individual customer between 2015 and 2020,
# and compare it to the average revenue of other customers within the same gender group.
%%sql
WITH net_revenue_base AS (SELECT customerkey, SUM(quantity * netprice / exchangerate) AS total_revenue
FROM sales
WHERE EXTRACT(YEAR FROM orderdate) BETWEEN 2015 AND 2020
GROUP BY 1),

net_revenue_gender AS (SELECT ntb.customerkey, c.gender, ntb.total_revenue
FROM net_revenue_base ntb
INNER JOIN customer c
ON ntb.customerkey = c.customerkey)

SELECT customerkey, gender, total_revenue,
AVG(total_revenue) OVER (PARTITION BY gender) AS avg_revenue_by_gender,
100 * total_revenue / AVG(total_revenue) OVER (PARTITION BY gender) AS revenue_vs_group
FROM net_revenue_gender
LIMIT 10
````

````python
# Store Order Ranking (3.3.1) - Problem
# Rank stores based on the total number of orders they received. This will help identify the highest-performing store locations by order volume.
%%sql

WITH store_orders AS (SELECT st.storecode, COUNT(*) AS total_orders
FROM sales s
LEFT JOIN store st
ON s.storekey = st.storekey
GROUP BY 1)

SELECT *, ROW_NUMBER() OVER(ORDER BY total_orders DESC) AS row_number,
RANK() OVER(ORDER BY total_orders DESC) AS rank,
DENSE_RANK() OVER(ORDER BY total_orders DESC) AS dense_rank
FROM store_orders
````

````python
# Customer Purchase Status (3.3.2) - Problem
# Identify which customers have made a recent purchase within the last year of available data. This helps distinguish active vs. inactive customers based on their most recent order.
%%sql
WITH sales_rn AS (SELECT ROW_NUMBER() OVER(PARTITION BY customerkey ORDER BY orderdate DESC) AS rn
,*
FROM sales)

SELECT customerkey,
orderdate AS last_purchase_date,
CASE WHEN orderdate BETWEEN
  (SELECT MAX(orderdate) FROM sales) - INTERVAL '1 year' AND (SELECT MAX(orderdate) FROM sales) THEN 'Recent Purchase'
  ELSE 'No Recent Purchase' END AS purchased_group
FROM sales_rn
WHERE rn = 1
LIMIT 10
````

````python
# Cohort LTV Ranking (3.3.3) - Problem
# Determine which customer cohorts (by year of first purchase) have the highest average lifetime value (LTV).
# This helps identify which acquisition years brought in the most valuable customers.
%%sql
WITH sales_cohort AS (
SELECT *,
EXTRACT(YEAR FROM MIN(orderdate) OVER (PARTITION BY customerkey)) AS cohort_year
FROM sales),

sales_revenue_cohort AS (
  SELECT customerkey,
  SUM(quantity * netprice * exchangerate) AS ltv,
  cohort_year
  FROM sales_cohort
  GROUP BY 1, 3
),

sales_ltv AS (SELECT customerkey, ltv, cohort_year,
AVG(ltv) OVER(PARTITION BY cohort_year) AS avg_cohort_ltv
FROM sales_revenue_cohort)

SELECT *,
DENSE_RANK() OVER(ORDER BY avg_cohort_ltv DESC) AS cohort_rank
FROM sales_ltv
````

````python
## Easy way to do it would be, but I though we need other columns and I didn't think we can use group by and window function in one query
%%sql
WITH yearly_cohort AS (
    SELECT
        customerkey,
        EXTRACT(YEAR FROM MIN(orderdate)) AS cohort_year,
        SUM(quantity * netprice * exchangerate) AS customer_ltv
    FROM sales
    GROUP BY
        customerkey
)
SELECT
    cohort_year,
    AVG(customer_ltv) AS avg_ltv,
    DENSE_RANK() OVER (ORDER BY AVG(customer_ltv) DESC) AS dense_rank
FROM yearly_cohort
GROUP BY
    cohort_year
````

But I though we can't use group by and window function in one query?

That's a common point of confusion, but your code is actually correct and demonstrates a valid pattern!

The key to understanding why this works is the order of operations in SQL:

* FROM / JOIN: Tables are joined.
* WHERE: Rows are filtered.
* GROUP BY: Rows are aggregated into groups.
* Aggregate Functions (e.g., SUM(), AVG() without OVER()): These are calculated for each group.
* HAVING: Groups are filtered.
* Window Functions (OVER()): These are calculated after the GROUP BY and aggregate functions have produced their results. Window functions operate on the result set of the GROUP BY clause, not the original rows.
* SELECT: Final columns are selected.
* DISTINCT: Duplicate rows are removed.
* ORDER BY: Results are sorted.
* LIMIT / OFFSET: Results are paginated.

In your query:

* The yearly_cohort CTE first groups by customerkey and calculates MIN(orderdate) and SUM(quantity * netprice * exchangerate). This is a standard GROUP BY aggregation.
* The outer SELECT statement then takes the results from yearly_cohort.
* It performs another GROUP BY cohort_year to calculate AVG(customer_ltv) for each cohort.
* Crucially, the DENSE_RANK() OVER (ORDER BY AVG(customer_ltv) DESC) window function is applied after the AVG(customer_ltv) has been computed for each cohort_year group. It operates on these aggregated results, ranking the cohort_year groups based on their average LTV.

So, while you can't typically use a window function directly within the same SELECT clause's GROUP BY as an aggregation target (e.g., GROUP BY DENSE_RANK()...), you can use window functions to process the results of aggregations, which is exactly what your query does. It's a very powerful and common pattern in SQL!

````python
# Monthly Cohort Analysis (3.3.4) - Problem
# Track how customer purchasing behavior changes over time by assigning each customer to a cohort year (based on the year of their first purchase),
# and then analyzing order activity on a monthly basis.
%%sql
WITH cohort_table AS (
    SELECT
        customerkey,
        EXTRACT(YEAR FROM MIN(orderdate)) AS cohort_year
    FROM sales
    GROUP BY 1),

totals_chort_month AS (
SELECT cohort_year,
TO_CHAR(orderdate, 'MM-YYYY') AS order_month,
COUNT(DISTINCT s.orderkey) AS total_orders,
COUNT(DISTINCT s.customerkey) AS total_customers
FROM sales s
LEFT JOIN cohort_table c
ON s.customerkey = c.customerkey
GROUP BY 1, 2)

SELECT *,
ROW_NUMBER() OVER(ORDER BY total_orders DESC) AS row_number,
RANK() OVER(ORDER BY total_orders DESC) AS rank,
DENSE_RANK() OVER(ORDER BY total_orders DESC) AS dense_rank
FROM totals_chort_month
LIMIT 10
````
